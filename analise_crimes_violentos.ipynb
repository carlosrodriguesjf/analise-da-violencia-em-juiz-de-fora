```python
# excluindo as colunas Unnamed,	Número REDS,		Natureza Principal Completa	Ano Fato	Mês Fato Resumido	Mês Numérico Fato	Data Fato	Dia da Semana Fato	Horário Fato	...	Logradouro Ocorrência - Tipo	Bairro	Bairro Não Cadastrado	Bairro - FATO FINAL	Bairro - FATO FINAL - Municipio	Município	Município - Código	UF - Sigla	RISP	RMBH
tabela = tabela.drop(columns=['Unnamed: 0','Número REDS','Ano Fato','Mês Numérico Fato','Logradouro Ocorrência - Tipo','Faixa 6 Horas Fato','Descrição Grupo Local Imediato','Bairro','Bairro Não Cadastrado','Município','UF - Sigla','Município - Código','Bairro - FATO FINAL - Municipio','RISP ','RMBH'])
tabela.sample(5)
```

```python
tabela.columns
```

```python
tabela.rename(columns={'Qtde Ocorrências':'numero_ocorrencias', 'Natureza Principal Completa':'crime',
       'Mês Fato Resumido':'mes',  'Data Fato':'data','Dia da Semana Fato':'dia_semana', 'Horário Fato':'horario',
       'Faixa 1 Hora Fato':'faixa_horaria_crime','Causa Presumida':'causa', 'Desc Longa Meio Utilizado':'meio_utilizado',
        'Descrição Local Imediato':'local', 'Bairro - FATO FINAL':'bairro'},inplace=True)
```

```python
tabela.columns
```

```python
# Convertendo colunas numéricas para o tipo inteiro
tabela['numero_ocorrencias'] = tabela['numero_ocorrencias'].astype(int)
tabela['data'] = tabela['data'].astype('datetime64[ns]')
```

```python
# criando a coluna ano 
tabela['ano'] = tabela['data'].dt.year
```

```python
# criando a coluna hora ocorrido 
tabela['hora_ocorrido'] = tabela['horario'].apply(lambda x: x.strip().split(':')[0])
```
