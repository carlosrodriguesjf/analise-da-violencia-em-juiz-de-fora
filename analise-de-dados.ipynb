```python
import pandas as pd
```

```python
tabela_alvos_furto1 = pd.read_csv('dados\Alvos - Furto - Jan 2015 a Dez 2017.csv', sep = ';', encoding_errors='ignore')
tabela_alvos_furto2 = pd.read_csv('dados\Alvos - Furto - Jan 2018 a Dez 2021.csv', sep = ';', encoding_errors='ignore')
tabela_alvos_furto3 = pd.read_csv('dados\Alvos - Furto - Jan 2022 a Jan 2025.csv', sep = ';', encoding_errors='ignore', dtype='str')
```

```python
tabela_alvos_furto = pd.concat([tabela_alvos_furto1, tabela_alvos_furto2, tabela_alvos_furto3], ignore_index=True)
```

```python
tabela_alvos_furto.shape
```

```python
tabela_alvos_furto.head()
```

```python
tabela_alvos_roubo = pd.read_csv('dados\Alvos - Roubo - Jan 2015 a Jan 2025.csv', sep = ';', encoding_errors='ignore')
tabela_alvos_roubo.shape
```

```python
tabela_alvos = pd.concat([tabela_alvos_furto, tabela_alvos_roubo], ignore_index=True)
```

```python
tabela_alvos.shape
```

```python
tabela_crimes_violentos1 = pd.read_csv('dados\Crimes Violentos - Jan 2012 a Dez 2021.csv', sep = ';', encoding_errors='ignore', dtype='str')
tabela_crimes_violentos2 = pd.read_csv('dados\Crimes Violentos - Jan 2022 a Jan 2025.csv', sep = ';', encoding_errors='ignore', dtype='str')
```

```python
tabela_crimes_violentos = pd.concat([tabela_crimes_violentos1, tabela_crimes_violentos2], ignore_index=True)
```

```python
tabela_crimes_violentos.shape
```

```python
tabela_lesao_corporal = pd.read_csv('dados\Lesão Corporal - Jan 2012 a Jan 2025.csv', sep = ';', encoding_errors='ignore', dtype='str')
```

```python
tabela_lesao_corporal.shape
```

```python
tabela_veiculos_furto = pd.read_csv('dados\Veiculos - Furto - Jan 2015 a Jan 2025.csv', sep = ';', encoding_errors='ignore',dtype = 'str')
tabela_veiculos_roubo = pd.read_csv('dados\Veiculos - Roubo - Jan 2015 a Jan 2025.csv', sep = ';', encoding_errors='ignore',dtype = 'str')
```

```python
tabela_veiculos = pd.concat([tabela_veiculos_furto, tabela_veiculos_roubo], ignore_index=True)
```

```python
tabela_veiculos.shape
```

```python
tabela_vitimas_homicidio = pd.read_csv('dados\Vítimas de Homicidio Consumado - Jan 2012 a Jan 2025.csv', sep = ';', encoding_errors='ignore',dtype = 'str')
```

```python
df_crimes = pd.concat([tabela_alvos, tabela_crimes_violentos, tabela_lesao_corporal,tabela_veiculos,tabela_vitimas_homicidio], ignore_index=True)
```
